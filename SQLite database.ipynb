```python
import sqlite3
```

```python
#connecting to sqlite
#connecting sqlite
conn = sqlite3.connect('INSTRUCTOR.db')
```

```python
#You can create Cursor object using the cursor() method of the Connection object/class
cursor_obj = conn.cursor()
```

```python
#drop table if already exits
cursor_obj.execute("DROP TABLE IF EXISTS INSTRUCTIOR")
```

```python
#creating table
table = """ create table IF NOT EXISTS INSTRUCTOR(
ID INTEGER PRIMARY KEY NOT NULL, 
FNAME VARCHAR(20),
 LNAME VARCHAR(20), 
 CITY VARCHAR(20), 
 CCODE CHAR(2)
 );"""
 
cursor_obj.execute(table)
print("TABLE IS READY")
```

```python
#We will start by inserting just the first row of data, i.e. for instructor Rav Ahuja
cursor_obj.execute(''' insert inTO INSTRUCTOR values(1, 'Rav', 'Ahuja', 'TORONTO', 'CA')''')
```

```python
#query data into table
select = '''SELECT * FROM INSTRUCTOR'''
cursor_obj.execute(select)
print('PRINT ALL THE DATA')
output_all = cursor_obj.fetchall()
for row_all in output_all:
    print(row_all)
```

```python
#fetch few rows from table
select = '''select * FROM INSTRUCTOR'''
cursor_obj.execute(select)
print("print all output")
output_all = cursor_obj.fetchmany(2)
for row_all in output_all:
    print(row_all)
```

```python
#fetch only fname
select = '''SELECT FNAME FROM INSTRUCTOR'''
cursor_obj.execute(select)
print("print all output")
output_all = cursor_obj.fetchall()
for fname in output_all:
    print(fname)
```

```python
#Bonus: now write and execute an update statement that changes the Rav's CITY to MOOSETOWN
update_city = '''UPDATE INSTRUCTOR
                SET CITY = 'MOOSETOWN'
                WHERE FNAME = "Rav"'''
cursor_obj.execute(update_city)
```

```python
select = '''SELECT * FROM INSTRUCTOR'''
cursor_obj.execute(select)
print("all done")
output_all = cursor_obj.fetchmany(2)
for row_all in output_all:
    print(row_all)
```
